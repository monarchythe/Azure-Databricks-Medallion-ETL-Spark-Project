### Dynamic Capabelities

```python
dbutils.widgets.text("file_name", "")
```

```python
p_file_name = dbutils.widgets.get("file_name")
```

### Reading data 

```python
df = spark.read.format("parquet").load("abfss://bronze@monarchazuredatalake.dfs.core.windows.net/orders")
df.display()
```

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format","parquet")\
        .option("cloudFiles.schemaLocation",f"abfss://bronze@monarchazuredatalake.dfs.core.windows.net/checkpoint_{p_file_name}")\
            .load(f"abfss://source@monarchazuredatalake.dfs.core.windows.net/{p_file_name}")
```

### write the data 

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
        .option("checkpointLocation",f"abfss://bronze@monarchazuredatalake.dfs.core.windows.net/checkpoint_{p_file_name}")\
            .option("path",f"abfss://bronze@monarchazuredatalake.dfs.core.windows.net/{p_file_name}")\
                .trigger(once=True)\
                    .start()
```
